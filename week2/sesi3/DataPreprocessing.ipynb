```python
path = "weight-height_outlier.csv"
df2 = pd.read_csv(path)
df2
```

Kode ini digunakan untuk membaca data dari file weight-height_outlier.csv. Data tersebut disimpan ke dalam df2 sehingga bisa dilihat dan diolah.

```python
import seaborn as sns
sns.boxplot(df2['Height'])
```

Kode ini digunakan untuk membuat boxplot dari data Height. Grafik ini digunakan untuk melihat penyebaran data dan menemukan data yang terlalu jauh dari data lainnya atau outlier.


```python
import numpy as np
#identifikasi Q1 Q3
q1, q3 = np.percentile(df2['Height'], [25, 75])
#Hitung IQR
iqr = q3 - q1
#cari lower and upper bounds
lower_bound = q1 - (1.5 * iqr)
upper_bound = q3 + (1.5 * iqr)
#hapus outliers
clean_data = df2[(df2['Height'] >= lower_bound)
                 & (df2['Height'] <= upper_bound)]

sns.boxplot(clean_data['Height'])
```

Kode ini digunakan untuk mencari dan menghapus data outlier pada kolom Height. Q1 dan Q3 digunakan untuk menghitung IQR, kemudian ditentukan batas bawah dan batas atas. Data yang berada di luar batas tersebut dihapus sehingga tersisa data yang lebih bersih.

```python
sns.boxplot(df2['Weight'])
```

Kode ini digunakan untuk membuat boxplot dari data Weight. Grafik ini digunakan untuk melihat penyebaran data dan menemukan data yang terlalu jauh dari data lainnya atau outlier.

```python
# Identifikasi Q1 Q3
q1, q3 = np.percentile(df2['Weight'], [25, 75])
# Hitung IQR
iqr = q3 - q1

# Cari Lower dan upper bounds
lower_bound = q1 - (1.5 * iqr)
upper_bound = q3 + (1.5 * iqr)
# Hapus outliers
clean_data = df2[(df2['Weight'] >= lower_bound)
& (df2['Weight'] <= upper_bound)]

sns.boxplot(clean_data['Weight' ])
```

Kode ini digunakan untuk mencari dan menghapus data outlier pada kolom Weight. Q1 dan Q3 digunakan untuk menghitung IQR, kemudian ditentukan batas bawah dan batas atas. Data di luar batas tersebut dihapus sehingga tersisa data yang lebih bersih.

```python
import matplotlib.pyplot as plt

# Membuat daftar kolom yang ingin dianalisis, yaitu "Weight" dan "Height".
columns_to_plot = ['Weight', 'Height' ]

# Memastikan kolom ada di dataset
columns_to_plot = [col for col in columns_to_plot if col in df2.columns]

fig, axs = plt.subplots(len(columns_to_plot), 1, dpi=95, figsize=(7, 5 * len(columns_to_plot)))

if len(columns_to_plot) == 1:
    axs = [axs]

for i, col in enumerate(columns_to_plot):
    axs[i].boxplot(df2[col], vert=False)
    axs[i].set_ylabel(col)

plt.show()
```

Kode ini digunakan untuk membuat boxplot dari kolom Weight dan Height. Grafik dibuat untuk melihat penyebaran data dan menemukan data yang memiliki nilai terlalu jauh atau outlier.

```python
def remove_outliers_sequentially(df2, columns):
    df_cleaned = df2.copy()
    for col in columns:
        q1, q3 = np.percentile(df_cleaned[col], [25, 75])
        iqr = q3 - q1
        lower_bound = q1 - (1.5 * iqr)
        upper_bound = q3 + (1.5 * iqr)
        df_cleaned = df_cleaned[(df_cleaned[col] >= lower_bound) & (df_cleaned[col])]
    return df_cleaned

# Terapkan fungsi untuk menghapus outliers dari "Height" dan "Weight"
clean_data = remove_outliers_sequentially(df2, ['Height', 'Weight'])

# Plot boxplots setelah menghapus outliers
columns_to_plot = ['Weight', 'Height']
fig, axs = plt.subplots(len(columns_to_plot), 1, dpi=95, figsize=(7, 5 * len(columns_to_plot)))

if len(columns_to_plot) == 1:
    axs = [axs ]

for i, col in enumerate(columns_to_plot):
    axs [i].boxplot (clean_data[col], vert=False)
    axs[i].set_ylabel(col)

plt.show()
```

Kode ini digunakan untuk menghapus outlier pada kolom Height dan Weight secara berurutan. Batas outlier dihitung menggunakan Q1, Q3, dan IQR, kemudian data yang berada di luar batas dihapus. Hasil akhirnya ditampilkan dalam bentuk boxplot.

```python
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler(feature_range=(0, 1))

# Memisahkan kolom numerik dan non-numerik
numeric_cols = clean_data.select_dtypes(include=[np.number]).columns
non_numeric_cols = clean_data.select_dtypes(exclude=[np.number]).columns

rescaled_numeric = scaler.fit_transform(clean_data[numeric_cols])
rescaled_numeric[ : ]
```

Kode ini digunakan untuk mengubah nilai pada kolom numerik ke rentang 0 sampai 1. Kolom angka dipisahkan dari kolom non-angka, kemudian data numerik diubah menggunakan MinMaxScaler sehingga nilainya menjadi lebih seragam.


```python
# Konversi kembali hasil scaling ke DataFrame
rescaled_df = pd.DataFrame(rescaled_numeric, columns=numeric_cols, index=clean_data.index)
# Gabungkan kembali dengan kolom non-numerik tanpa mengubah jumlah baris
final_df = pd.concat([clean_data[non_numeric_cols], rescaled_df], axis=1)
final_df
```

Kode ini mengubah hasil scaling menjadi DataFrame, lalu menggabungkannya kembali dengan kolom non-numerik. Hasil akhirnya disimpan dalam final_df sehingga semua data tetap menjadi satu tabel.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

path = "diabetes.csv"

df_diabetes = pd.read_csv(path)

# Pilih 3 kolom untuk deteksi outlier
columns_to_plot = ['Glucose', 'BloodPressure', 'BMI']

# Membuat boxplot
fig, axs = plt.subplots(len(columns_to_plot), 1, dpi=95, figsize=(7, 5 * len(columns_to_plot)))

for i, col in enumerate(columns_to_plot):
    axs[i].boxplot(df_diabetes[col], vert=False)
    axs[i].set_ylabel(col)

plt.show()
```

```python
clean_diabetes = df_diabetes.copy()

for col in columns_to_plot:
    q1, q3 = np.percentile(clean_diabetes[col], [25, 75])
    iqr = q3 - q1

    lower_bound = q1 - (1.5 * iqr)
    upper_bound = q3 + (1.5 * iqr)

    clean_diabetes = clean_diabetes[
        (clean_diabetes[col] >= lower_bound) &
        (clean_diabetes[col] <= upper_bound)
    ]

clean_diabetes
```

Kode ini digunakan untuk mendeteksi dan menghapus outlier pada kolom Glucose, BloodPressure, dan BMI. Outlier dicari menggunakan boxplot dan metode IQR, kemudian data yang berada di luar batas ditentukan sebagai outlier dan dihapus.
