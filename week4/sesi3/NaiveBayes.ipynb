```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris

iris = load_iris()
```

Kode di atas digunakan untuk mengimpor library NumPy dan Pandas, kemudian memuat dataset Iris dari Scikit-learn untuk digunakan dalam analisis data.

```python
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df["Species"] = iris.target_names[iris.target]

X = df.drop(columns=["Species"])
y = df["Species"]
print(X)
```

Kode ini digunakan untuk mengubah dataset Iris menjadi DataFrame Pandas dan menambahkan kolom Species. Kemudian, data dipisahkan menjadi X sebagai fitur dan y sebagai target. print(X) digunakan untuk menampilkan seluruh data fitur.

```python
y
```

Menampilkan isi y yaitu data spesies bunga Iris yang menjadi target atau label, terdiri dari setosa, versicolor, dan virginica.

```python
from sklearn.preprocessing import LabelEncoder
lab = LabelEncoder()
y = lab.fit_transform(y)
y
```

Kode digunakan untuk mengubah label spesies bunga Iris dari teks menjadi angka menggunakan LabelEncoder. Spesies setosa, versicolor, dan virginica diubah menjadi angka 0, 1, dan 2, kemudian hasilnya ditampilkan.

```python
from sklearn.model_selection import train_test_split
X_train, X_test = train_test_split(X, test_size=0.2, random_state=0)
y_train, y_test = train_test_split(y, test_size=0.2, random_state=0)
```

Kode digunakan untuk membagi data menjadi data training dan data testing. Sebanyak 80% data digunakan untuk training dan 20% untuk testing. random_state=0 membuat pembagian data tetap sama setiap kali kode dijalankan.

```python
from sklearn.naive_bayes import GaussianNB

model = GaussianNB()
NB_model = model.fit(X_train, y_train)
```

Kode ini digunakan untuk membuat dan melatih model Naive Bayes dengan metode GaussianNB. Model dilatih menggunakan data X_train dan y_train, kemudian hasil model disimpan dalam NB_model untuk digunakan dalam proses prediksi.

```python
result = NB_model.predict(X_test)
result
```

Kode ini digunakan untuk melakukan prediksi menggunakan model Naive Bayes yang sudah dilatih sebelumnya. NB_model.predict(X_test) memprediksi kelas dari data pengujian X_test, kemudian hasil prediksi disimpan ke dalam variabel result. Baris result digunakan untuk menampilkan hasil prediksi tersebut.

```python
from sklearn.metrics import accuracy_score
print('Accuracy of data testing: ', accuracy_score(y_test, result)*100)
```

Kode ini digunakan untuk menghitung akurasi model Naive Bayes pada data testing. accuracy_score(y_test, result) membandingkan hasil sebenarnya pada y_test dengan hasil prediksi pada result. Hasilnya dikalikan 100 sehingga akurasi ditampilkan dalam bentuk persentase.

```python
input_data = [[5.1, 3.5, 1.4, 0.2]]
result = NB_model.predict(input_data)
prediction_label = lab.inverse_transform(result)
print('Hasil prediksi untuk input baru adalah:', prediction_label)
```

```python
input_data = pd.DataFrame(np.array([[5.1, 3.5, 1.4, 0.2]]), columns=iris.feature_names)
result = NB_model.predict(input_data)
prediction_label = lab.inverse_transform(result)
print('Hasil prediksi untuk input baru adalah:', prediction_label)
```

Kode ini digunakan untuk memprediksi jenis bunga Iris dari data baru. input_data berisi nilai panjang dan lebar sepal serta panjang dan lebar petal dari bunga yang ingin diprediksi. Data tersebut kemudian diprediksi menggunakan model Naive Bayes dan hasilnya masih berupa angka, sehingga lab.inverse_transform(result) digunakan untuk mengubahnya kembali menjadi nama spesies seperti setosa, versicolor, atau virginica.

```python
input_tugas = pd.DataFrame(np.array([[7.3, 3.8, 4.6, 2.3]]), columns=iris.feature_names)
result_tugas = NB_model.predict(input_tugas)
prediction_label = lab.inverse_transform(result_tugas)
print('Hasil prediksi untuk data [7.3, 3.8, 4.6, 2.3] adalah:', prediction_label)
```

Kode ini digunakan untuk memprediksi spesies bunga Iris dari data baru [7.3, 3.8, 4.6, 2.3]. Data tersebut dibuat menjadi DataFrame dengan nama fitur Iris, lalu diprediksi menggunakan model Naive Bayes. Hasil prediksi yang masih berupa angka diubah kembali menjadi nama spesies menggunakan inverse_transform(), kemudian hasilnya ditampilkan.
