```python
import numpy as np

#matriks 5x5
matriks = np.random.randint(0, 50, size=(5, 5))
print("Matriks 5x5:")
print(matriks)

#nilai maksimum, minimum, rata-rata
nilai_max = np.max(matriks)
nilai_min = np.min(matriks)
nilai_rata = np.mean(matriks)

print(f"Nilai Maksimum : {nilai_max}")
print(f"Nilai Minimum  : {nilai_min}")
print(f"Nilai Rata-rata: {nilai_rata}")
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-2, 2, 200)
y = np.linspace(-2, 2, 200)
X, Y = np.meshgrid(x, y)

#soal
z = X**6 + 3*X**4*Y**2 - 3*X**4 + 3*X**2*Y**4 - X**2*Y**3 - 6*X**2*Y**2 + 3*X**2 + Y**6 - 3*Y**4 + 3*Y**2

plt.contour(X, Y, z, [1])
plt.title("Grafik 2D")
plt.xlabel("x")
plt.ylabel("y")
plt.grid()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-3, 3, 50)
y = np.linspace(-3, 3, 50)
X, Y = np.meshgrid(x, y)

#soal
Z = (5*Y/4 - np.abs(X))**2 + X**2

ax = plt.axes(projection='3d')
ax.plot_surface(X, Y, Z)

plt.title("Grafik 3D")
plt.show()
```
